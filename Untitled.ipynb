```python
df['Startup_Age'] = 2026 - df['Founding_Year']
```

```python
df['Startup_Age'].head(20)
```

```python
df['Startup_Age'].describe()
```

```python
bins = [0,5,10,15,20,100]
labels = ['0-5 years', '6-10 years', 
          '11 - 15 years', '16-20 years', 
          '20+ years']
df['Startup_Age_Group'] = pd.cut(
    df['Startup_Age'], 
    bins=bins,
    labels=labels
)
```

```python
df['Startup_Age_Group'].value_counts().sort_index()
```

```python
df['Startup_Age_Group'].value_counts().sort_index().plot(kind= 'bar',figsize= (8,5))
plt.title("Startup Distribution by Age")
plt.xlabel('Startup Age')
plt.ylabel('Number of Startups')
plt.xticks(rotation=0 )
plt.show()
```

```python
df['Total_Funding_USD_Millions'].describe()
```

```python
df.groupby('Funding_Stage').agg(Startups = ('Company_ID','count'), 
 Avg_Funding = ('Total_Funding_USD_Millions','median'), 
 Median_Funding = ('Total_Funding_USD_Millions','mean'), 
 Avg_Valuation = ('Valuation_USD_Millions','mean'),  
Median_Valuation =('Valuation_USD_Millions','median')).sort_values('Avg_Funding',ascending=False)
```

```python
stage_funding = df.groupby('Funding_Stage')['Total_Funding_USD_Millions'].median()
stage_funding.sort_values(ascending = False).plot(kind='bar',figsize=(9,5))
plt.title('Median Funding by Funding Stage')
plt.xlabel('Funding Stage')
plt.ylabel('Median Funding (USD Millions)')
plt.xticks(rotation=0)
plt.show()
```
